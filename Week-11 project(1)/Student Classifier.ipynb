```python

# OOP Class
class StudentClassifier:

    def __init__(self, dataframe):
        self.dataframe = dataframe
        self.__error_message = "Error: Student age is greater than 25"

    # Getter method
    def get_error_message(self):
        return self.__error_message

    # Classification method
    def classifier(self):

        # Pirates
        pirates = self.dataframe[
            (self.dataframe['Age'] > 14) &
            (self.dataframe['Age'] < 18)
        ]

        pirates.to_csv('The_Pirates.csv', index=False)

        print("\nTHE PIRATES")
        print(pirates)

        # Yankees
        yankees = self.dataframe[
            (self.dataframe['Age'] > 18) &
            (self.dataframe['Age'] < 22)
        ]

        yankees.to_csv('The_Yankees.csv', index=False)

        print("\nTHE YANKEES")
        print(yankees)

        # Bulls
        bulls = self.dataframe[
            (self.dataframe['Age'] > 22) &
            (self.dataframe['Age'] < 25)
        ]

        bulls.to_csv('The_Bulls.csv', index=False)

        print("\nTHE BULLS")
        print(bulls)

        # Students above 25
        above_25 = self.dataframe[
            self.dataframe['Age'] > 25
        ]

        print("\nSTUDENTS ABOVE 25")

        for index, row in above_25.iterrows():
            print(f"{row['Name']} --> {self.get_error_message()}")


# Create object
student_data = StudentClassifier(df)

# Run method
student_data.classifier()
```
