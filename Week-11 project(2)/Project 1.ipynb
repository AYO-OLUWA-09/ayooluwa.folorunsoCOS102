```python
from abc import ABC, abstractmethod

# Basd class
class coup_de_escriva(ABC):

    @abstractmethod
    def Fan_page(self):
        pass

    # Child classes
class FC_Cirok(coup_de_escriva):

    def Fan_page(self):
        print("Welcome to FC_Cirok")

class Madiba_FC(coup_de_escriva):
    def Fan_page(self):
        print("Welcome to Madiba_FC")

class Blue_Jay_Fc(coup_de_escriva):
    def Fan_page(self):
        print("Welcome to Blue_Jay_FC")

class TSG_Walker(coup_de_escriva):

    def Fan_page(self):
        print("Welcome to TSG_Walker")

## Getting user input
club = input("Enter your club name: ")

if club == "FC_Cirok":
    fan = FC_Cirok()

elif club == "Madiba_FC":
    fan =  Madiba_FC()
    
elif club == "Blue_Jay_Fc":
    fan = Blue_Jay_Fc()

elif club == "TSG_Walker":
    fan = TSG_Walker()

else:
    print("Club not found")
    exit()

# Invoke Method
fan.Fan_Page()
```
