```python
from abc import ABC, abstractmethod
```

```python

#Base Class
class External_Vendors(ABC):

    @abstractmethod
    def menu(self):
        pass


# Cooperative Cafeteria
class Cooperative_Hostel(External_Vendors):

    def menu(self):

        print("\nCooperative Cafeteria Menu")
        print("Jollof Rice and Stew - N200")
        print("White Rice and Stew - N200")
        print("Fried Rice - N200")
        print("Salad - N100")
        print("Plantain - N100")


# Faith Hostel Cafeteria
class Faith_Hostel(External_Vendors):

    def menu(self):

        print("\nFaith Hostel Cafeteria Menu")
        print("Fried Rice - N400")
        print("White Rice and Stew - N400")
        print("Jollof Rice - N400")
        print("Beans - N200")
        print("Chicken - N1000")


# Student Centre Cafeteria
class Student_Center(External_Vendors):

    def menu(self):

        print("\nStudent Centre Cafeteria Menu")
        print("Chicken Fried Rice - N800")
        print("Pomo Sauce - N300")
        print("Spaghetti Jollof - N500")
        print("Amala/Ewedu - N500")
        print("Semo with Eforiro Soup - N500")


# User Input
vendor = input("Enter Vendor Name: ")

# Instantiate Object
if vendor == "Cooperative Hostel":
    food = Cooperative_Hostel()

elif vendor == "Faith Hostel":
    food = Faith_Hostel()

elif vendor == "Student Center":
    food = Student_Center()

else:
    print("Vendor not found")
    exit()

# Display Menu
food.menu()
```
