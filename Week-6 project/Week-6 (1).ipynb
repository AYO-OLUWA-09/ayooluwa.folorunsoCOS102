```python
import tkinter as tk
from tkinter import messagebox
import random
from datetime import datetime
```

```python
# Creating the main window
root = tk.Tk()
root.title("Employees Login-Task Assigner System")
root.geometry("400x300")

# Create username label and entry
label1 = tk.Label(root, text="Username:")
label1.pack()
#Entry box
name_entry = tk.Entry(root)
name_entry.pack()
# Result label
result_label = tk.Label(root, text="")
result_label.pack()


# The submit button function
def submit():
    name = name_entry.get()

   # using the class created
    if emp_system.check_employee(name):
        task = emp_system.assign_task()
        status, time = emp_system.take_attendance(name, task)

        if status:
            # first time today
            result_label.config(text=" ATTENDANCE RECORDED\n\nEmployee: " + name + 
                               "\nTime: " + time +
                               "\nTask: " + task)
            
       
    else:
         result_label.config(text="✗ ACCESS DENIED\n\n" + name + " is not an employee")

       
    
## Employee Class
class Employee:
    def __init__(self):
        # list of employees
         self.employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", 
                         "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James", 
                         "Michell Taiwo", "Abraham Jones", "Nicole Anide", "Kosi Korso", 
                         "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"]

         self.tasks = ["Loading", "Transporting", "Reviewing Orders", "Customer Service", "Delivering Items"]
         self.attendance = {}
    def check_employee(self, name):
        # To chck if the name of the employee is in the list using the if statemen
        if name in self.employees:
            return True
        else:
            return False

    def assign_task(self):
       import random
       task_index = random.randint(0, len(self.tasks) - 1)
       return self.tasks[task_index]

    def save_to_file(self, name, task, time):
        from datetime import datetime
        try:
            with open("attendance_log.txt", "a") as file:
                file.write(f"{datetime.now().strftime('%Y-%m-%d %H:%M:%S')} | {name} | Task: {task}\n")
            return True
        except:
            return False
    def take_attendance(self, name, task):
        from datetime import datetime
        today = datetime.now().strftime("%Y-%m-%d")
        current_time = datetime.now().strftime("%H:%M:%S")

        if name in self.attendance and self.attendance[name]["date"] == today:
            return False, self.attendance[name]["time"]

        self.attendance[name] = {
            "date": today,
            "time": current_time,
            "task": task
        }
        return True, current_time
        
# create submit button
submit_button = tk.Button(root, text="Identity Check", command=submit)
submit_button.pack()

# Employee object
emp_system = Employee()
task = emp_system.assign_task()
root.mainloop()
```
