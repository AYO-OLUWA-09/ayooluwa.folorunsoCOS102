```python
import tkinter as tk
from tkinter import messagebox
import random
from datetime import datetime
```

```python
# STEP 1: IMPORT THE TOOLS WE NEED
# Import the tkinter library and give it a shorter name 'tk' for easier typing
#import tkinter as tk
# Import the messagebox module specifically to show popup error messages
#from tkinter import messagebox


# STEP 2: CREATE A BLUEPRINT FOR OUR PROGRAM USING A CLASS
# 'class' is like a template or blueprint. 'DeliveryApp' is the name of our program blueprint
class DeliveryApp:
    # This is the CONSTRUCTOR function - it runs automatically when we create the program
    # 'self' represents the program itself, 'root' is the main window we will create
    def __init__(self, root):
        # Store the main window inside our program so we can use it anywhere
        self.root = root
        
        # Set the title that appears at the top of the window
        self.root.title("Delivery Charge Calculator")
        
        # Set the window size: 400 pixels wide and 300 pixels tall
        self.root.geometry("400x300")
        
        # Make the window NOT resizable - False for width, False for height
        # This prevents users from stretching the window
        self.root.resizable(False, False)
        
        # STEP 3: CREATE VARIABLES TO STORE USER INPUT
        # Create a special variable to store the location (PAU or Epe)
        # StringVar() creates a text variable that automatically updates the screen
        # value="PAU" sets PAU as the default selected option
        self.location_var = tk.StringVar(value="PAU")
        
        # Create another special variable to store the weight the user will type
        # Starts empty because user hasn't typed anything yet
        self.weight_var = tk.StringVar()
        
        # STEP 4: CALL THE FUNCTION THAT BUILDS ALL THE BUTTONS AND LABELS
        # This runs the create_widgets function which we will define next
        self.create_widgets()
    
    
    # STEP 5: CREATE THE FUNCTION THAT BUILDS EVERYTHING THE USER SEES ON SCREEN
    def create_widgets(self):
        # --- CREATE THE TITLE LABEL ---
        # Label creates a piece of text on the screen
        # self.root is the main window, text is what to display
        # font sets the style, size, and boldness of the text
        title_label = tk.Label(self.root, text="Delivery Charge Calculator", font=("Arial", 16, "bold"))
        # pack places the label in the window
        # pady=10 adds 10 pixels of space above and below the label
        title_label.pack(pady=10)
        
        
        # --- CREATE THE LOCATION SELECTION BOX WITH RADIO BUTTONS ---
        # LabelFrame creates a box with a border and title to group related items
        # text="Select Location" is the title of this box
        # padx=10 adds 10 pixels of space inside the box on left and right
        # pady=10 adds 10 pixels of space inside the box on top and bottom
        location_frame = tk.LabelFrame(self.root, text="Select Location", padx=10, pady=10)
        # pack places this frame in the window
        # pady=10 adds 10 pixels space above and below
        # fill="x" makes the frame stretch horizontally across the window
        # padx=20 adds 20 pixels space on left and right outside the frame
        location_frame.pack(pady=10, fill="x", padx=20)
        
        # Create the first radio button for PAU location
        # Radiobutton creates a circular button that can be selected
        # location_frame is the box that will contain this radio button
        # text="PAU" is the label next to the radio button
        # variable=self.location_var links this button to our location storage variable
        # value="PAU" means when selected, store "PAU" in location_var
        tk.Radiobutton(location_frame, text="PAU", variable=self.location_var, value="PAU").pack(anchor="w")
        # anchor="w" aligns the radio button to the west (left side)
        
        # Create the second radio button for Epe location
        # Same as above but with text and value set to "Epe"
        tk.Radiobutton(location_frame, text="Epe", variable=self.location_var, value="Epe").pack(anchor="w")
        
        
        # --- CREATE THE WEIGHT INPUT BOX ---
        # Create another labeled frame for the weight input section
        weight_frame = tk.LabelFrame(self.root, text="Package Weight (kg)", padx=10, pady=10)
        # Place this frame below the previous one with spacing and stretching
        weight_frame.pack(pady=10, fill="x", padx=20)
        
        # Entry creates a blank box where users can type text
        # weight_frame is the box that will contain this entry field
        # textvariable=self.weight_var links this entry box to our weight storage variable
        # width=20 sets how wide the typing box is (20 characters)
        tk.Entry(weight_frame, textvariable=self.weight_var, width=20).pack(pady=5)
        # pady=5 adds 5 pixels of space above and below the entry box
        # .pack() places it inside the weight_frame
        
        
        # --- CREATE THE CALCULATE BUTTON ---
        # Button creates a clickable button on the screen
        # self.root is the main window that will contain this button
        # text="Calculate Charge" is what appears on the button
        # command=self.calculate_charge tells the program which function to run when clicked
        # bg="blue" sets the button's background color to blue
        # fg="white" sets the button's text color to white
        # font sets the button text style and size
        calc_button = tk.Button(self.root, text="Calculate Charge", command=self.calculate_charge, bg="blue", fg="white", font=("Arial", 12))
        # Place the button in the window with 10 pixels space above and below
        calc_button.pack(pady=10)
        
        
        # --- CREATE THE RESULT DISPLAY LABEL ---
        # Create a label that will show the calculated delivery charge
        # Initially text is empty (starts blank)
        # fg="green" sets the text color to green when it appears
        self.result_label = tk.Label(self.root, text="", font=("Arial", 12, "bold"), fg="green")
        # Place this label with 10 pixels space above and below
        self.result_label.pack(pady=10)
    
    
    # STEP 6: CREATE THE FUNCTION THAT CALCULATES THE DELIVERY CHARGE
    # This function runs when the user clicks the Calculate button
    def calculate_charge(self):
        # Get the location the user selected from the storage variable
        # .get() retrieves whatever value is stored (either "PAU" or "Epe")
        location = self.location_var.get()
        
        # Get the weight the user typed from the storage variable
        # This comes as text (string) because that's what the user typed
        weight_str = self.weight_var.get()
        
        
        # STEP 7: CHECK IF USER ENTERED A WEIGHT (VALIDATION)
        # if not weight_str means: if the weight box is empty
        if not weight_str:
            # Show an error popup window
            # "Input Error" is the title of the error window
            # "Please enter the package weight." is the error message
            messagebox.showerror("Input Error", "Please enter the package weight.")
            # return stops the function immediately - don't continue with calculation
            return
        
        
        # STEP 8: CHECK IF THE WEIGHT IS A VALID NUMBER
        # try: means "try to do this, but if there's an error, don't crash"
        try:
            # Convert the text the user typed into a decimal number
            # Example: "10.5" becomes 10.5, "7" becomes 7.0
            weight = float(weight_str)
        # except ValueError: runs if the conversion fails (user typed letters or symbols)
        except ValueError:
            # Show an error popup telling user to enter a number
            messagebox.showerror("Input Error", "Weight must be a number.")
            # Stop the function
            return
        
        
        # STEP 9: APPLY THE PRICING RULES BASED ON LOCATION AND WEIGHT
        # Check which location the user selected
        if location == "PAU":
            # User selected PAU location
            # Check if weight is 10kg or more
            if weight >= 10:
                # Weight is 10kg or above - charge N2000
                charge = 2000
            else:
                # Weight is less than 10kg - charge N1500
                charge = 1500
        elif location == "Epe":
            # User selected Epe location
            # Check if weight is 10kg or more
            if weight >= 10:
                # Weight is 10kg or above - charge N5000
                charge = 5000
            else:
                # Weight is less than 10kg - charge N4000
                charge = 4000
        else:
            # This should never happen since only PAU or Epe can be selected
            # But just in case, set charge to None (no value)
            charge = None
        
        
        # STEP 10: DISPLAY THE RESULT ON THE SCREEN
        # .config() changes the text of the result label
        # f"Delivery Charge: N{charge}" creates a formatted string
        # Example: if charge is 2000, it becomes "Delivery Charge: N2000"
        self.result_label.config(text=f"Delivery Charge: N{charge}")


# STEP 11: START THE PROGRAM (THIS RUNS WHEN YOU EXECUTE THE FILE)
# if __name__ == "__main__": means "only run this code if this file is run directly"
# This prevents the code from running if this file is imported into another program
if __name__ == "__main__":
    # Create the main window - this is like creating an empty stage
    # tk.Tk() creates the main application window
    root = tk.Tk()
    
    # Create an object (instance) of our DeliveryApp class
    # This actually builds our program inside the window
    # The root window is passed to the class so it knows where to put everything
    app = DeliveryApp(root)
    
    # Start the program's main loop - this keeps the window open and responsive
    # Without this, the window would appear and immediately close
    # This loop waits for user actions like clicks and key presses
    root.mainloop()
```
